````python
import tensorflow as tf
import json
````

````python
# Load facial expression images
# Folder structure:
# facial-expression/train/angry/
# facial-expression/train/disgust/
# facial-expression/train/fear/
# facial-expression/train/happy/
# facial-expression/train/neutral/
# facial-expression/train/sad/
# facial-expression/train/surprise/

data = tf.keras.utils.image_dataset_from_directory(
    "FER_56_per_class/train",
    image_size=(64, 64),
    batch_size=8,
    label_mode="int"
)

print("Classes:", data.class_names)
````

````python
# CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(64, 64, 3)),
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(len(data.class_names), activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
````

````python
# Train
model.fit(data, epochs=50)
model.summary()
````

````python
# Prediction
# Replace 'sample.jpg' with an image that exists in your test folder.
img = tf.keras.utils.load_img(
    "FER_56_per_class/test/angry\im2.png",
    target_size=(64, 64)
)
x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = model.predict(x)
print("Predicted expression:", data.class_names[p.argmax()])
print("Confidence:", round(float(p.max()) * 100, 2), "%")
````

````python
# Save model and class names
model.save("facial_expression.keras")

with open("classes.json", "w") as f:
    json.dump(data.class_names, f)
````

````python
# Load saved model and predict a new image
with open("classes.json", "r") as f:
    class_names = json.load(f)

m = tf.keras.models.load_model("facial_expression.keras")

img = tf.keras.utils.load_img(
    "FER_56_per_class/test/angry/im2.png",
    target_size=(64, 64)
)
x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = m.predict(x)
print(p)
print("Predicted expression:", class_names[p.argmax()])
````
